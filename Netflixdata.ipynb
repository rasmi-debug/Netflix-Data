```python
#yesto garepaxi paxi gayerw malai jun sukai column ko max , min ,% tha paunu xa vane call garne pugxa
def info(df,cols,labels):

    edges =[df[cols].describe()['min'],
           df[cols].describe()["25%"],
           df[cols].describe()["50%"],
           df[cols].describe()["75%"],
           df[cols].describe()["max"]]

           
    df[cols] = pd.cut(df[cols], edges , labels = labels, duplicates = 'drop')     
    return df
labels = ['not_popular','below_avg','average','popular']    
```

```python
info(df,'vote_average',labels)
df['vote_average'].unique()
```

```python
df.groupby("Title")["Popularity"].sum().sindex().plot(
    kind = 'bar'
    figsize = (8,8)
)
plt.xlabels("title")
plt.ylabels("popularity")
plt.title("dfjkdf")
plt.show()
```

```python
#bar ko laagi kind, fig, ani x and y axis hunxa
df.groupby("Title")["Popularity"].sum().plot(
    kind='bar',
    figsize=(8, 8),
   
)
plt.xlabel("Title")
plt.ylabel("Popurarity")
plt.title("popularity as according to title")
plt.show()
```

```python
#pie chart ko lagi kind, fig, autopct and x,y axis hudaina
df.groupby("Title")["Popularity"].sum().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(8, 8)
)

plt.title("Out of 100% , popularity of each movie compare to one another")

plt.show()
```

```python

#line graph ma kind and fig size .sum.sort_index.plot


df.groupby('release_date')['Popularity'].sum().sort_index().plot(
    kind='line',
    figsize=(10,5)
)

plt.xlabel("Release Date")   # X-axis = date
plt.ylabel("Popularity")     # Y-axis = value
plt.title("Popularity Over Time")

plt.show()
```
